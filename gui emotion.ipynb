```python
import tkinter as tk
from tkinter import filedialog
from tkinter import *
from PIL import Image, ImageTk
import numpy as np
from keras.models import load_model
import librosa

# Loading the pre-trained emotion detection model
model = load_model('emotion_detection_model.h5')

# Initialize Tkinter GUI window
top = tk.Tk()
top.geometry('800x600')
top.title('Emotion Detector')
top.configure(background='#CDCDCD')

# Create Labels to show the predicted emotion
label1 = Label(top, background="#CDCDCD", font=('arial', 15, "bold"))
sign_image = Label(top)

# Define the emotion labels based on the indices provided
emotion_labels = {
    0: 'oaf_angry', 1: 'oaf_disgust', 2: 'oaf_fear', 3: 'oaf_happy',
    4: 'oaf_neutral', 5: 'oaf_pleasant_surprise', 6: 'oaf_sad',
    7: 'yaf_angry', 8: 'yaf_disgust', 9: 'yaf_fear', 10: 'yaf_happy',
    11: 'yaf_neutral', 12: 'yaf_pleasant_surprised', 13: 'yaf_sad'
}

# Function to process the audio file and predict the emotion
def Detect(file_path):
    try:
        # Load the audio file and extract features (MFCCs)
        audio_data, sr = librosa.load(file_path, sr=None)
        mfccs = librosa.feature.mfcc(y=audio_data, sr=sr, n_mfcc=40)
        mfccs = np.mean(mfccs.T, axis=0)

        # Reshaping to match the model input
        mfccs = np.expand_dims(mfccs, axis=0)

        # Normalize the input
        mfccs = mfccs / np.max(np.abs(mfccs))

        # Predict the emotion using the pre-trained model
        pred = model.predict(mfccs)
        
        # Get the emotion with the highest probability
        predicted_index = np.argmax(pred)
        predicted_emotion = emotion_labels[predicted_index]
        print(f"Predicted Emotion: {predicted_emotion}")
        
        # Update the label with the prediction result
        label1.configure(foreground="#011638", text=predicted_emotion)

    except Exception as e:
        print("Error in detection:", e)

# Function to show the Detect button after file is uploaded
def show_Detect_button(file_path):
    Detect_b = Button(top, text="Detect Emotion", command=lambda: Detect(file_path), padx=10, pady=5)
    Detect_b.configure(background="#364156", foreground='white', font=('arial', 10, 'bold'))
    Detect_b.place(relx=0.79, rely=0.46)

# Function to upload an audio file
def upload_audio():
    try:
        # Open a file dialog to select an audio file
        file_path = filedialog.askopenfilename(filetypes=[("Audio Files", "*.wav *.mp3")])
        
        # Display the selected audio file's name in the GUI
        sign_image.configure(text="Audio File Selected: " + file_path.split("/")[-1])
        label1.configure(text='')

        # Show the Detect Emotion button
        show_Detect_button(file_path)

    except Exception as e:
        print("Error:", e)

# Upload button to trigger the file selection
upload = Button(top, text="Upload an Audio File", command=upload_audio, padx=10, pady=5)
upload.configure(background="#364156", foreground='white', font=('arial', 10, 'bold'))
upload.pack(side='bottom', pady=50)

# Place the audio file label and emotion prediction label sections
sign_image.pack(side='bottom', expand=True)
label1.pack(side="bottom", expand=True)

# Title of the application
heading = Label(top, text="Emotion Detection from Audio", pady=20, font=('arial', 20, "bold"))
heading.configure(background="#CDCDCD", foreground="#364156")
heading.pack()

# Run the GUI
top.mainloop()
```
